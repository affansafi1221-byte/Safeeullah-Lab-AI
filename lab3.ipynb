```python
from google import genai

client = genai.Client(api_key="")
MODEL = "models/gemini-3.5-flash"

def website_auditor(user_input):
    prompt = f"""
You are WebGuard AI, a website audit agent.

Analyze the website information below and identify:
- SEO problems
- Accessibility problems
- Usability problems
- Technical issues

For each issue give:
Issue, Severity (Low/Medium/High), Recommendation.

Website information:
{user_input}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text

website = input("Enter website information: ")
print(website_auditor(website))
```
